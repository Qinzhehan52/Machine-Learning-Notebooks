## 逻辑回归的平方误差？
<img align="left" src="./images/C1_W3_SqErrorVsLogistic.png"     style=" width:400px; padding: 10px; " > 回想一下，对于**线性**回归，我们使用了**平方误差代价函数**：
单变量平方误差代价的方程为：
  $$J(w,b) = \frac{1}{2m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)})^2 \tag{1}$$ 
 
其中
  $$f_{w,b}(x^{(i)}) = wx^{(i)} + b \tag{2}$$


回想一下，平方误差代价具有良好的性质，即跟随代价的导数会导致最小值。

```python
soup_bowl()
```

这个代价函数对于线性回归效果很好，自然也会考虑将其用于逻辑回归。然而，正如上面的幻灯片所指出的，$f_{wb}(x)$ 现在有一个非线性分量，即 Sigmoid 函数：$f_{w,b}(x^{(i)}) = sigmoid(wx^{(i)} + b )$。让我们在之前实验的例子上尝试平方误差代价，现在包括 Sigmoid。

这是我们的训练数据：

```python
x_train = np.array([0., 1, 2, 3, 4, 5],dtype=np.longdouble)
y_train = np.array([0,  0, 0, 1, 1, 1],dtype=np.longdouble)
plt_simple_example(x_train, y_train)
```

现在，让我们使用*平方误差代价*获取代价的表面图：
  $$J(w,b) = \frac{1}{2m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)})^2 $$ 
 
其中
  $$f_{w,b}(x^{(i)}) = sigmoid(wx^{(i)} + b )$$


```python
plt.close('all')
plt_logistic_squared_error(x_train,y_train)
plt.show()
```

虽然这产生了一个非常有趣的图，但上面的表面远不如线性回归的“汤碗”那么光滑！

逻辑回归需要一个更适合其非线性性质的代价函数。这从损失函数开始。这将在下面描述。

## 逻辑损失函数

<img align="left" src="./images/C1_W3_LogisticLoss_b.png"     style=" width:250px; padding: 2px; " >

逻辑回归使用一种更适合分类任务的损失函数，其中目标是 0 或 1，而不是任何数字。

>**定义说明：**   在本课程中，使用以下定义：
**损失（Loss）** 是衡量单个示例与其目标值差异的指标，而
**代价（Cost）** 是衡量训练集上损失的指标


定义如下：
* $loss(f_{\mathbf{w},b}(\mathbf{x}^{(i)}), y^{(i)})$ 是单个数据点的代价，即：

\begin{equation}
  loss(f_{\mathbf{w},b}(\mathbf{x}^{(i)}), y^{(i)}) = \begin{cases}
    - \log\left(f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) & \text{如果 $y^{(i)}=1$}\\
    - \log \left( 1 - f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) & \text{如果 $y^{(i)}=0$}
  \end{cases}
\end{equation}


*  $f_{\mathbf{w},b}(\mathbf{x}^{(i)})$ 是模型的预测，而 $y^{(i)}$ 是目标值。

*  $f_{\mathbf{w},b}(\mathbf{x}^{(i)}) = g(\mathbf{w} \cdot\mathbf{x}^{(i)}+b)$ 其中函数 $g$ 是 Sigmoid 函数。

该损失函数的定义特征是它使用两条单独的曲线。一条用于目标为零或 ($y=0$) 的情况，另一条用于目标为一 ($y=1$) 的情况。结合起来，这些曲线提供了对损失函数有用的行为，即当预测与目标匹配时为零，并且随着预测与目标偏离而迅速增加值。考虑下面的曲线：

```python
plt_two_logistic_loss_curves()
```

结合起来，这些曲线类似于平方误差损失的二次曲线。注意，x 轴是 $f_{\mathbf{w},b}$，它是 Sigmoid 的输出。Sigmoid 输出严格在 0 和 1 之间。

上面的损失函数可以重写以便更容易实现。
    $$loss(f_{\mathbf{w},b}(\mathbf{x}^{(i)}), y^{(i)}) = (-y^{(i)} \log\left(f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) - \left( 1 - y^{(i)}\right) \log \left( 1 - f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right)$$
  
这是一个看起来相当可怕的方程。当你考虑到 $y^{(i)}$ 只能有两个值，0 和 1 时，它就不那么令人生畏了。然后可以将方程分为两部分：
当 $ y^{(i)} = 0$ 时，左边的项被消除：
$$
\begin{align}
loss(f_{\mathbf{w},b}(\mathbf{x}^{(i)}), 0) &= (-(0) \log\left(f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) - \left( 1 - 0\right) \log \left( 1 - f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) \\
&= -\log \left( 1 - f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right)
\end{align}
$$
当 $ y^{(i)} = 1$ 时，右边的项被消除：
$$
\begin{align}
  loss(f_{\mathbf{w},b}(\mathbf{x}^{(i)}), 1) &=  (-(1) \log\left(f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) - \left( 1 - 1\right) \log \left( 1 - f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right)\\
  &=  -\log\left(f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right)
\end{align}
$$

好了，有了这个新的逻辑损失函数，就可以产生一个包含所有示例损失的代价函数。这将是下一个实验的主题。现在，让我们来看看我们上面考虑的简单示例的代价与参数曲线：

```python
plt.close('all')
cst = plt_logistic_cost(x_train,y_train)
```

这条曲线非常适合梯度下降！它没有平稳区、局部最小值或不连续点。注意，它不像平方误差那样是一个碗。绘制了代价和代价的对数，以阐明曲线在代价很小时具有斜率并继续下降的事实。提醒：你可以使用鼠标旋转上面的图。
